```python
import numpy as np

def rotate(points, angle_degrees, pivot=(0, 0)):
    angle = np.radians(angle_degrees)
    px, py = pivot
    rot_matrix = np.array([
        [np.cos(angle), -np.sin(angle)],
        [np.sin(angle), np.cos(angle)]
    ])

    rotated_points = []
    for x, y in points:
        translated = np.array([x - px, y - py])
        rotated = rot_matrix.dot(translated)
        rotated_points.append((rotated[0] + px, rotated[1] + py))
    return rotated_points

# Example
if __name__ == "__main__":
    square = [(0, 0), (2, 0), (2, 2), (0, 2)]
    rotated = rotate(square, 45, pivot=(1, 1))
    print("Rotated points:", rotated)
```
