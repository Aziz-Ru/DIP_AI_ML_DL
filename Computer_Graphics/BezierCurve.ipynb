```python
import matplotlib.pyplot as plt
import numpy as np
from math import comb

def bezier_curve(control_points, num_points=100):
    n = len(control_points) - 1
    control_points = np.array(control_points)
    t_values = np.linspace(0, 1, num_points)
    curve = np.zeros((num_points, 2))

    for i, t in enumerate(t_values):
        point = np.zeros(2)
        for k in range(n + 1):
            bernstein = comb(n, k) * (t ** k) * ((1 - t) ** (n - k))
            point += bernstein * control_points[k]
        curve[i] = point
    return curve

def draw_bezier(control_points):
    curve = bezier_curve(control_points)
    cx, cy = zip(*control_points)
    plt.plot(cx, cy, 'ro--', label="Control Polygon")
    plt.plot(curve[:, 0], curve[:, 1], 'b-', label="Bezier Curve")
    plt.legend()
    plt.title("Bezier Curve")
    plt.show()

# Example
if __name__ == "__main__":
    control_points = [(0, 0), (2, 5), (5, 5), (7, 0)]
    draw_bezier(control_points)
```
