```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def breasham_line(x1,y1,x2,y2):
    dx = abs(x2 - x1)
    dy = abs(y2 - y1)
    sx = 1 if x1 < x2 else -1
    sy = 1 if y1 < y2 else -1
    m = dy/dx if dx != 0 else 0
    px, py = [], []
    if m<1:
        x,y = x1,y1
        p = 2*dy - dx
        while x != x2:
            px.append(x)
            py.append(y)
            if p<0:
                x+= sx
                p+=2*dy
            else:
                x+= sx
                y+= sy
                p+= 2*dy - 2*dx
    else:
        x,y = x1,y1
        p = 2*dx - dy
        while y != y2:
            px.append(x)
            py.append(y)
            if p<0:
                y+= sy
                p+=2*dx
            else:
                x+= sx
                y+= sy
                p+= 2*dx - 2*dy
    return px, py

```

```python

px,py = breasham_line(0,0,5,7)
plt.plot(px,py)
plt.title("Bresenham's Line Algorithm")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid()
plt.show()
```
