```python
import numpy as np

def scale(points, sx, sy, pivot=(0, 0)):
    px, py = pivot
    scaled_points = []
    for x, y in points:
        new_x = px + (x - px) * sx
        new_y = py + (y - py) * sy
        scaled_points.append((new_x, new_y))
    return scaled_points

# Example
if __name__ == "__main__":
    square = [(0, 0), (2, 0), (2, 2), (0, 2)]
    scaled = scale(square, 2, 2, pivot=(1, 1))
    print("Scaled points:", scaled)
```
