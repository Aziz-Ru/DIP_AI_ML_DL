```python
import matplotlib.pyplot as plt
```

```python
def dda_line(x1,y1,x2,y2):
  px,py =[],[]
  dx = abs(x2-x1)
  dy = abs(y2-y1)
  steps = max(dx,dy)
  xinc = dx/steps
  yinc = dy/steps
  x,y = x1,y1
  for _ in range(steps+1):
    px.append(round(x))
    py.append(round(y))
    x += xinc
    y += yinc
  return px,py

```

```python
px,py = dda_line(2,3,10,7)
plt.plot(px,py)
plt.title("DDA Line Drawing Algorithm")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid()
plt.show()
```
