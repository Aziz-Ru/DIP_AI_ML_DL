```python
def clip_polygon(polygon, xmin, ymin, xmax, ymax):
    def inside(p, edge):
        x, y = p
        if edge == 'left':
            return x >= xmin
        if edge == 'right':
            return x <= xmax
        if edge == 'bottom':
            return y >= ymin
        if edge == 'top':
            return y <= ymax

    def intersect(p1, p2, edge):
        x1, y1 = p1
        x2, y2 = p2
        if edge in ('left', 'right'):
            x_edge = xmin if edge == 'left' else xmax
            y = y1 + (y2 - y1) * (x_edge - x1) / (x2 - x1)
            return (x_edge, y)
        else:
            y_edge = ymin if edge == 'bottom' else ymax
            x = x1 + (x2 - x1) * (y_edge - y1) / (y2 - y1)
            return (x, y_edge)

    for edge in ('left', 'right', 'bottom', 'top'):
        new_polygon = []
        n = len(polygon)
        for i in range(n):
            curr = polygon[i]
            prev = polygon[i - 1]
            curr_in = inside(curr, edge)
            prev_in = inside(prev, edge)

            if curr_in:
                if not prev_in:
                    new_polygon.append(intersect(prev, curr, edge))
                new_polygon.append(curr)
            elif prev_in:
                new_polygon.append(intersect(prev, curr, edge))
        polygon = new_polygon
    return polygon

# Example
if __name__ == "__main__":
    subject_polygon = [(1, 1), (4, 1), (4, 4), (1, 4)]
    clipped = clip_polygon(subject_polygon, 2, 2, 6, 6)
    print("Clipped polygon:", clipped)
```
