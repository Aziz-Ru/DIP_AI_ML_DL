```python
import matplotlib.pyplot as plt
import numpy as np

def koch_curve(p1, p2, depth):
    if depth == 0:
        return [p1, p2]

    p1 = np.array(p1, dtype=float)
    p2 = np.array(p2, dtype=float)

    # Divide segment into three equal parts
    a = p1 + (p2 - p1) / 3
    b = p1 + 2 * (p2 - p1) / 3

    # Compute the peak point of the equilateral triangle
    angle = np.pi / 3  # 60 degrees
    direction = b - a
    rot_matrix = np.array([
        [np.cos(angle), -np.sin(angle)],
        [np.sin(angle), np.cos(angle)]
    ])
    peak = a + rot_matrix.dot(direction)

    points = []
    points += koch_curve(p1, a, depth - 1)[:-1]
    points += koch_curve(a, peak, depth - 1)[:-1]
    points += koch_curve(peak, b, depth - 1)[:-1]
    points += koch_curve(b, p2, depth - 1)
    return points

def draw_koch(depth=4):
    p1 = (0, 0)
    p2 = (10, 0)
    points = koch_curve(p1, p2, depth)
    xs, ys = zip(*points)
    plt.plot(xs, ys, 'b-')
    plt.axis('equal')
    plt.title(f"Koch Curve (depth={depth})")
    plt.show()

# Example
if __name__ == "__main__":
    draw_koch(depth=4)
```
