```python
import numpy as np

# Create Sample Dataset
# create a dataset with 10 samples and 2 features

X = np.array([
    [2.5, 2.4],
    [0.5, 0.7],
    [2.2, 2.9],
    [1.9, 2.2],
    [3.1, 3.0],
    [2.3, 2.7],
    [2.0, 1.6],
    [1.0, 1.1],
    [1.5, 1.6],
    [1.1, 0.9]
])

print("Original Data:")
print(X)
```

```python
rows = len(X)
cols = len(X[0])
mean = []
for j in range(cols):
    column_sum = 0
    for i in range(rows):
        column_sum += X[i][j]
    mean.append(column_sum / rows)

print("Mean Vector:")
print(mean)

centered_data = []
for i in range(rows):
    centered_row = []
    for j in range(cols):
        centered_row.append(X[i][j] - mean[j])
    centered_data.append(centered_row)

print("Centered Data:")
print(centered_data)


cov_matrix = []

for i in range(cols):
    row = []

    for j in range(cols):
        cov = 0

        for k in range(rows):
            cov += centered_data[k][i] * centered_data[k][j]

        cov = cov / (rows - 1)
        row.append(cov)

    cov_matrix.append(row)

print("Covariance Matrix:")
for row in cov_matrix:
    print(row)

```

```python
mean = np.mean(X, axis=0)

print("Mean Vector:")
print(mean)
```

```python
X_centered = X - mean

print("Centered Data:")
print(X_centered)
```

```python
cov_matrix = np.cov(X_centered.T)

print("Covariance Matrix:")
print(cov_matrix)
```

```python
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

print("Eigenvalues:")
print(eigenvalues)

print("Eigenvectors:")
print(eigenvectors)
```

```python
sorted_index = np.argsort(eigenvalues)[::-1]

eigenvalues = eigenvalues[sorted_index]
eigenvectors = eigenvectors[:, sorted_index]

print("Sorted Eigenvalues:")
print(eigenvalues)

print("Sorted Eigenvectors:")
print(eigenvectors)
```

```python
k = 1

principal_components = eigenvectors[:, :k]

print("Principal Component:")
print(principal_components)
```

```python
X_pca = X_centered @ principal_components

print("Data after PCA:")
print(X_pca)
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=k)
X_pca = pca.fit_transform(X)

print("PCA Result:")
print(X_pca)
```

```python
explained_variance = eigenvalues / np.sum(eigenvalues)

print("Explained variance ratio:")
print(explained_variance)

print("Percentage:")
print(explained_variance * 100)
```
